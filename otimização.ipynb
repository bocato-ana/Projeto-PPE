```python
df = pd.read_csv('data_2024.csv')
df.columns.tolist()
```

```python
# Força a exibição de toda a lista de nomes no terminal da célula
print(list(df.columns))
```

```python
dicionario_colunas = {
        '1.a.1_faixa_idade': 'Idade',
        '1.b_genero': 'Gênero',
        '1.c_cor/raca/etnia': 'Cor/Etnia',
        '1.d_pcd': 'PCD',
        '1.e_experiencia_profissional_prejudicada': 'Experiência prejudicada',
        '2.g_nivel': 'Nivel',
        '2.h_faixa_salarial': 'Salário',
        '2.j_tempo_de_experiencia_em_ti': 'Tempo de experiencia',
        
    }
```

```python
df = df[list(dicionario_colunas.keys())]
```

```python
df.info(memory_usage="deep")
```

```python
df['1.c_cor/raca/etnia'].unique()
```

```python
df['1.a.1_faixa_idade'] = df['1.a.1_faixa_idade'].map({'17-21': 0, '22-24': 1, '25-29':2, '30-34': 3, '35-39': 4, '40-44': 5, '45-49': 6, '50-54': 7, '55+': 8}).astype('Int8')
df['1.b_genero'] = df['1.b_genero'].map({'Masculino': 0, 'Feminino': 1, 'Prefiro não informar': 0, 'Outro': 0}).astype('bool')
```

```python
df['1.a.1_faixa_idade'].unique()
```

nessa ultima aí coloquei o prefiro não infromar com base nas nossas analises, onde eles são homens que ganham bem.... absurdo

```python
etnia = df['1.c_cor/raca/etnia'].str.get_dummies(sep=', ')
df = pd.concat([df, etnia], axis=1)
df
```

```python
# Coloque o nome das colunas que quer excluir dentro da lista
df = df.drop(columns=['1.c_cor/raca/etnia', 'Indígena', 'Prefiro não informar', 'Outra', '2.j_tempo_de_experiencia_em_ti'])
```

```python
df
```

```python
df['2.g_nivel'].unique()
```

```python
# 1. Mapeia apenas os textos reais que existem na coluna
df['2.g_nivel'] = df['2.g_nivel'].map({
    'Júnior': 0, 
    'Pleno': 1, 
    'Sênior': 2
})

# 2. Trata os vazios verdadeiros preenchendo com -1 e força a conversão final
df['2.g_nivel'] = df['2.g_nivel'].fillna(-1).astype(int)

df
```

```python
# 1. Criamos o dicionário apontando várias faixas para o mesmo grupo (0, 1, 2, 3 ou 4)
dicionario_salario_agrupado = {
    # GRUPO 0: Até R$ 3.000
    'Menos de R$ 1.000/mês': 0,
    'de R$ 1.001/mês a R$ 2.000/mês': 0,
    'de R$ 2.001/mês a R$ 3.000/mês': 0,

    # GRUPO 1: De R$ 3.001 a R$ 6.000
    'de R$ 3.001/mês a R$ 4.000/mês': 1,
    'de R$ 4.001/mês a R$ 6.000/mês': 1,

    # GRUPO 2: De R$ 6.001 a R$ 12.000
    'de R$ 6.001/mês a R$ 8.000/mês': 2,
    'de R$ 8.001/mês a R$ 12.000/mês': 2,

    # GRUPO 3: De R$ 12.001 a R$ 20.000
    'de R$ 12.001/mês a R$ 16.000/mês': 3,
    'de R$ 16.001/mês a R$ 20.000/mês': 3,

    # GRUPO 4: Acima de R$ 20.000
    'de R$ 20.001/mês a R$ 25.000/mês': 4,
    'de R$ 25.001/mês a R$ 30.000/mês': 4,
    'de R$ 30.001/mês a R$ 40.000/mês': 4,
    'Acima de R$ 40.001/mês': 4
}

# 2. Aplicamos a substituição na sua coluna
df['2.h_faixa_salarial'] = df['2.h_faixa_salarial'].map(dicionario_salario_agrupado)

# 3. Tratamos os NaN (espaços em branco) usando aquela mesma lógica do -1
df['2.h_faixa_salarial'] = df['2.h_faixa_salarial'].fillna(-1).astype(int)
```

```python
df
```

```python
# Procura a palavra 'Sim' dentro do texto de cada linha
# O na=False garante que quem deixou a pergunta em branco (NaN) vire False automaticamente
df['1.e_experiencia_profissional_prejudicada'] = df['1.e_experiencia_profissional_prejudicada'].str.contains('Sim', na=False)
```

```python
df
```

```python
df['1.e_experiencia_profissional_prejudicada'].unique()
```

```python
# Procura explicitamente pela palavra 'Sim'. 
# Se achar, vira True. Se não achar (o que inclui o texto 'Não' ou vazio), vira False.
df['1.d_pcd'] = df['1.d_pcd'].str.contains('Sim', na=False, case=False)
```
